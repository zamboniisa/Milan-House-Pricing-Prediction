```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error
import xgboost as xgb
from sklearn.model_selection import GridSearchCV, KFold
```

```python
# Load the data
train_data = pd.read_csv('Encoded_Train.csv')
# train_data.drop('w', axis=1, inplace=True)
test_data = pd.read_csv('Encoded_Test.csv')
```

```python
# Define features and target
X_train = train_data.drop('y', axis=1)
y_train = train_data['y']
```

```python
# Normalize the feature data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(test_data)
```

```python
# parameters choice
parameters = { 'objective': ['reg:tweedie'], 'enable_categorical': [True], 'random_state':[88], 'max_depth': [5], 
               'min_child_weight': [0], 'subsample': [0.66], 'colsample_bytree': [0.235], 'learning_rate': [0.02], 
               'n_estimators': [16000], 'lambda': [0.0], 'alpha': [0.2], 'gamma': [0.4], 'validate_parameters': [True],
               'tree_method': ['hist'], 'max_leaves': [200], 'booster': ['gbtree']}

model = xgb.XGBRegressor(**parameters)

kfold = KFold(n_splits = 10, shuffle = True, random_state = 88)
grid_search = GridSearchCV(estimator = model, param_grid = parameters, cv = kfold, scoring = 'neg_mean_absolute_error', n_jobs = -1, verbose = 2)
grid_search.fit(X_train_scaled, y_train)

y_pred = grid_search.predict(X_test_scaled)
```

```python
# Save predictions to a text file
with open('predictions.txt', 'w') as f:
    for pred in y_pred:
        f.write(f"{pred}\n")
```
